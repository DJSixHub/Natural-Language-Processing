```python
import json
import re
from pathlib import Path

ruta_json = Path(r"E:/Escuela/NLP/Natural-Language-Processing/Jsons_Scrappeados/recetas.json")
ruta_salida = ruta_json.parent / "recetas_normalizado.json"

# Frases a eliminar
frases_irrelevantes = [
    r"sube la foto de.*", 
    r"si te ha gustado.*", 
    r"no te pierdas.*",
    r"\(por comensal\)",
    r"\(por unidad\)"
]

def normalizar_texto(texto):
    texto = texto.lower().strip()
    # quitar espacios extra
    texto = re.sub(r"\s+", " ", texto)
    # eliminar frases irrelevantes
    for frase in frases_irrelevantes:
        texto = re.sub(frase, "", texto)
    texto = texto.strip()
    return texto

with open(ruta_json, "r", encoding="utf-8") as f:
    recetas = json.load(f)

recetas_normalizadas = []

for receta in recetas:
    # Normalizar nombre quitando "Receta de" y pasando a minúsculas
    nombre = receta.get("Nombre", "").strip()
    nombre = re.sub(r"^receta de\s+", "", nombre, flags=re.I)
    nombre = nombre.lower()
    
    # Normalizar instrucciones
    instrucciones = receta.get("Instrucciones", [])
    instrucciones_norm = [normalizar_texto(inst) for inst in instrucciones if normalizar_texto(inst)]
    
    # Normalizar ingredientes
    ingredientes = receta.get("Ingredientes", [])
    ingredientes_norm = [normalizar_texto(ing) for ing in ingredientes if normalizar_texto(ing)]
    
    # Copiar otros campos sin modificar
    recetas_normalizadas.append({
        "Nombre": nombre,
        "Comensales": receta.get("Comensales", ""),
        "Dificultad": receta.get("Dificultad", ""),
        "Tiempo": receta.get("Tiempo", ""),
        "Ingredientes": ingredientes_norm,
        "Instrucciones": instrucciones_norm
    })

# Guardar JSON normalizado
with open(ruta_salida, "w", encoding="utf-8") as f:
    json.dump(recetas_normalizadas, f, ensure_ascii=False, indent=2)

print(f"JSON normalizado guardado en: {ruta_salida}")
```

```python
import json
import re
from pathlib import Path

ruta_json_utensilios = Path(r"E:/Escuela/NLP/Natural-Language-Processing/Jsons_Scrappeados/utensilios.json")
ruta_salida_utensilios = ruta_json_utensilios.parent / "utensilios_normalizado.json"

def normalizar_texto(texto):
    if not texto:
        return ""
    texto = texto.lower().strip()
    texto = re.sub(r"\s+", " ", texto)
    return texto

with open(ruta_json_utensilios, "r", encoding="utf-8") as f:
    utensilios = json.load(f)

utensilios_normalizados = []

for utensilio in utensilios:
    nombre = utensilio.get("nombre", "")
    nombre_norm = normalizar_texto(nombre)
    
    utensilios_normalizados.append({
        "nombre": nombre_norm,
        "url": utensilio.get("url", None)
    })

# Guardar JSON normalizado
with open(ruta_salida_utensilios, "w", encoding="utf-8") as f:
    json.dump(utensilios_normalizados, f, ensure_ascii=False, indent=2)

print(f"JSON de utensilios normalizado guardado en: {ruta_salida_utensilios}")
```
